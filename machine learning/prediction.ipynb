```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import pandas as pd
with open("train.csv","r") as file:
    data = pd.read_csv(file)
    print(data.select_dtypes(include="int").columns)
```

```python
y = data.LotArea
X = data.select_dtypes(include="int64").drop(columns=["Id","LotArea",'MoSold', 'YrSold','1stFlrSF', '2ndFlrSF'])
model = RandomForestRegressor(random_state=1)
model.fit(X,y)
print(data.LotArea)
print(model.predict(X.head()))
pre = model.predict(X)
print(mean_absolute_error(y,pre))
```

```python
with open("test.csv","r")as realfile:
    realdata = pd.read_csv(realfile)
    Y = realdata.LotArea
    x = realdata.select_dtypes(include="int64").drop(columns=["Id","LotArea",'MoSold', 'YrSold','1stFlrSF', '2ndFlrSF'])
    pred = model.predict(x)
    print(mean_absolute_error(Y,pred))
```
